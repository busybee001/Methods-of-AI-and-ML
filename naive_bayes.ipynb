```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn import model_selection
from sklearn.naive_bayes import GaussianNB

from utilities import visualize_classifier
```

```python
X = np.array([[2.1, 1.3],
              [1.8, 2.5],
              [3.6, 5.1],
              [4.2, 4.9],
              [7.3, 2.1],
              [8.1, 1.8]])

Y = np.array([0, 0, 1, 1, 2, 2])
```

```python
classifier = GaussianNB()

classifier.fit(X, Y)

visualize_classifier(classifier, X, Y)
```

```python
test_values = [[8.2, 3.1]]

predicted = classifier.predict(test_values)

print(predicted)
```

```python
visualize_classifier(classifier, X, Y)

plt.scatter(8.2, 3.1, marker='x', color='red', s=200)

plt.show()
```

```python
X_train, X_test, Y_train, Y_test = model_selection.train_test_split(
    X, Y, test_size=0.2, random_state=5
)
```

```python
classifier_new = GaussianNB()

classifier_new.fit(X_train, Y_train)
```

```python
Y_pred = classifier_new.predict(X_test)

visualize_classifier(classifier_new, X_test, Y_test)
```

```python
accuracy = 100.0 * (Y_test == Y_pred).sum() / X_test.shape[0]

print("Accuracy of the classifier =", round(accuracy, 2), "%")
```

```python
from sklearn.model_selection import cross_val_score
```

```python
accuracy = cross_val_score(classifier, X, Y, scoring='accuracy', cv=2)
print("Accuracy: " + str(round(100*accuracy.mean(), 2)) + "%")
```

```python
precision = cross_val_score(classifier, X, Y, scoring='precision_weighted', cv=2)
print("Precision: " + str(round(100*precision.mean(), 2)) + "%")
```

```python
recall = cross_val_score(classifier, X, Y, scoring='recall_weighted', cv=2)
print("Recall: " + str(round(100*recall.mean(), 2)) + "%")
```
