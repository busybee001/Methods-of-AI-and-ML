```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
```

```python
num_points = 1200

m = 0.2
c = 0.5

x_data = np.random.normal(0.0, 0.8, num_points)
noise = np.random.normal(0.0, 0.04, num_points)

y_data = m * x_data + c + noise
```

```python
plt.figure(figsize=(6, 5))
plt.plot(x_data, y_data, "ro", markersize=3)
plt.title("Input data")
plt.xlabel("Dimension 0")
plt.ylabel("Dimension 1")
plt.show()
```

```python
W = np.random.uniform(-1.0, 1.0)
b = 0.0

learning_rate = 0.5
num_iterations = 10
loss_values = []
```

```python
for step in range(num_iterations):
    y_pred = W * x_data + b
    error = y_pred - y_data
    loss = np.mean(error ** 2)

    dW = 2 * np.mean(error * x_data)
    db = 2 * np.mean(error)

    W = W - learning_rate * dW
    b = b - learning_rate * db

    loss_values.append(loss)

    print("ITERATION", step + 1)
    print("W =", W)
    print("b =", b)
    print("loss =", loss)

    plt.figure(figsize=(6, 5))
    plt.plot(x_data, y_data, "ro", markersize=3)
    plt.plot(x_data, W * x_data + b)
    plt.xlabel("Dimension 0")
    plt.ylabel("Dimension 1")
    plt.title("Iteration " + str(step + 1) + " of " + str(num_iterations))
    plt.show()
```

```python
plt.figure(figsize=(6, 5))
plt.plot(range(1, num_iterations + 1), loss_values, marker="o")
plt.title("Training Loss Progress")
plt.xlabel("Iteration")
plt.ylabel("Loss")
plt.grid(True)
plt.show()
```
