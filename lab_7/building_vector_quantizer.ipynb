```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

```python
np.random.seed(42)

data1 = np.random.randn(100, 2) + [2, 2]
data2 = np.random.randn(100, 2) + [7, 7]
data3 = np.random.randn(100, 2) + [2, 7]
data4 = np.random.randn(100, 2) + [7, 2]

X = np.vstack((data1, data2, data3, data4))
```

```python
plt.figure(figsize=(7,5))
plt.scatter(X[:,0], X[:,1])
plt.title("Input Data")
plt.xlabel("Dimension 1")
plt.ylabel("Dimension 2")
plt.show()
```

```python
kmeans = KMeans(
    n_clusters=4,
    random_state=42,
    n_init=10
)

kmeans.fit(X)

print("Cluster Centers:")
print(kmeans.cluster_centers_)
```

```python
plt.figure(figsize=(7,5))

plt.scatter(
    X[:,0],
    X[:,1],
    c=kmeans.labels_,
    cmap='viridis'
)

plt.scatter(
    kmeans.cluster_centers_[:,0],
    kmeans.cluster_centers_[:,1],
    marker='X',
    s=300
)

plt.title("Vector Quantization Result")
plt.xlabel("Dimension 1")
plt.ylabel("Dimension 2")
plt.show()
```

```python
print("Number of clusters:", kmeans.n_clusters)
print("Inertia:", kmeans.inertia_)
print("Cluster Centers:")
print(kmeans.cluster_centers_)
```
