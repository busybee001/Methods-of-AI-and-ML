```python
from kanren import Relation, facts, run, var
```

```python
# Create relations
parent = Relation()
male = Relation()
female = Relation()
```

```python
# Define parent relationships
facts(parent,
      ("John", "David"),
      ("John", "Emma"),
      ("Michael", "John"),
      ("Sophia", "John"),
      ("David", "Chris"))
```

```python
# Define males
facts(male,
      ("Michael",),
      ("John",),
      ("David",),
      ("Chris",))
```

```python
# Define females
facts(female,
      ("Sophia",),
      ("Emma",))
```

```python
# Create variable
x = var()
```

```python
# Find children of John
print("Children of John:")
print(run(0, x, parent("John", x)))
```

```python
# Find parents of John
print("Parents of John:")
print(run(0, x, parent(x, "John")))
```

```python
# Find male members
print("Male members:")
print(run(0, x, male(x)))
```

```python
# Find female members
print("Female members:")
print(run(0, x, female(x)))
```

```python
from kanren import conde
```

```python
# Define father relationship
def father(x, y):
    return conde((parent(x, y), male(x)))
```

```python
# Define mother relationship
def mother(x, y):
    return conde((parent(x, y), female(x)))
```

```python
# Define grandparent relationship
z = var()

def grandparent(x, y):
    return conde((parent(x, z), parent(z, y)))
```

```python
# Add more family data
facts(parent,
      ("David", "William"),
      ("Emma", "Adam"),
      ("Chris", "Wayne"))
```

```python
facts(male,
      ("William",),
      ("Adam",),
      ("Wayne",))
```

```python
# John's children
name = 'John'

output = run(0, x, father(name, x))

print("List of " + name + "'s children:")

for item in output:
    print(item)
```

```python
# William's mother
name = 'William'

output = run(0, x, mother(x, name))

print(name + "'s mother:")

for item in output:
    print(item)
```

```python
# Adam's parents
name = 'Adam'

output = run(0, x, parent(x, name))

print("List of " + name + "'s parents:")

for item in output:
    print(item)
```

```python
# Wayne's grandparents
name = 'Wayne'

output = run(0, x, grandparent(x, name))

print("List of " + name + "'s grandparents:")

for item in output:
    print(item)
```
