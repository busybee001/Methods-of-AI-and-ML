```python
import itertools as it
from sympy.ntheory.generate import prime, isprime
```

```python
def check_prime(number):
    """
    This function checks whether a number is prime.
    """
    return isprime(number)
```

```python
# Input list from the textbook example
numbers = (23, 4, 27, 17, 13, 10, 21, 29, 3, 32, 11, 19)
```

```python
# Find prime numbers from the given list
prime_numbers = []

for number in numbers:
    if check_prime(number):
        prime_numbers.append(number)

print("Prime numbers in the list:")
print(prime_numbers)
```

```python
# Print first 7 prime numbers
first_seven_primes = []

for i in range(1, 8):
    first_seven_primes.append(prime(i))

print("First 7 prime numbers:")
print(first_seven_primes)
```
