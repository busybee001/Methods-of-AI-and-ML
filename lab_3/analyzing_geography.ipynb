```python
from kanren import run, var, membero
```

```python
# Define neighboring states
neighbors = (
    ("Arizona", "California"),
    ("Arizona", "Nevada"),
    ("Arizona", "Utah"),
    ("Arizona", "New Mexico"),
    ("California", "Nevada"),
    ("Nevada", "Utah"),
    ("Utah", "Colorado"),
    ("Colorado", "New Mexico"),
    ("New Mexico", "Arizona")
)
```

```python
# Create logic variables
x = var()
y = var()
```

```python
# Find all neighbor pairs
print("List of neighboring states:")

output = run(0, (x, y), membero((x, y), neighbors))

for item in output:
    print(item)
```

```python
# Find neighbors of Arizona
state = "Arizona"

output = run(0, y, membero((state, y), neighbors))

print("Neighbors of " + state + ":")

for item in output:
    print(item)
```
