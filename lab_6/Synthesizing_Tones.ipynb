```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.io.wavfile import write

# Function to generate tone
def generate_tone(freq, duration, amplitude=1.0, sampling_freq=44100):
    time_axis = np.linspace(0, duration, int(duration * sampling_freq))
    signal = amplitude * np.sin(2 * np.pi * freq * time_axis)
    return signal

# Parameters
sampling_freq = 44100
duration = 2

# Generate tones
tone_1 = generate_tone(440, duration, 0.6, sampling_freq)   # A note
tone_2 = generate_tone(660, duration, 0.6, sampling_freq)   # E note

# Combine tones
combined_signal = tone_1 + tone_2

# Normalize signal
combined_signal = combined_signal / np.max(np.abs(combined_signal))

# Save audio file
write(
    "synthesized_tone.wav",
    sampling_freq,
    (combined_signal * 32767).astype(np.int16)
)

print("Synthesized tone saved as synthesized_tone.wav")

# Plot synthesized tone
plt.figure(figsize=(10, 5))
plt.plot(combined_signal[:1000], color="black")
plt.title("Synthesized Tone")
plt.xlabel("Sample")
plt.ylabel("Amplitude")
plt.show()
```
