```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.io import wavfile

# Generate sample audio signal
duration = 3
sampling_freq = 44100
frequency = 440

time = np.linspace(0, duration, int(sampling_freq * duration))

audio_signal = np.sin(2 * np.pi * frequency * time)

# Save audio file
wavfile.write(
    "sample_audio.wav",
    sampling_freq,
    (audio_signal * 32767).astype(np.int16)
)

# Read audio file
sampling_freq, signal = wavfile.read("sample_audio.wav")

print("Sampling frequency:", sampling_freq)
print("Signal shape:", signal.shape)
print("Duration:", round(signal.shape[0] / sampling_freq, 2), "seconds")

# Plot audio signal
plt.figure(figsize=(10, 5))
plt.plot(signal[:2000])
plt.title("Audio Signal")
plt.xlabel("Sample")
plt.ylabel("Amplitude")
plt.show()
```
