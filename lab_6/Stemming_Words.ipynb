```python
import nltk
nltk.download('punkt')

from nltk.stem.porter import PorterStemmer

stemmer = PorterStemmer()

words = [
    "table",
    "probably",
    "wolves",
    "playing",
    "is",
    "dog",
    "the",
    "beaches",
    "grounded",
    "dreamt",
    "envision"
]

print("Original Word --> Stemmed Word\n")

for word in words:
    print(f"{word} --> {stemmer.stem(word)}")
```
