```python
import nltk

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')

from nltk.tokenize import RegexpTokenizer
from nltk.corpus import stopwords
from nltk.stem.snowball import SnowballStemmer
from gensim import models, corpora

# Input data with two topics: mathematics and history
data = [
    "Mathematics includes numbers formulas equations and calculations.",
    "Algebra and geometry are important branches of mathematics.",
    "Students solve math problems using formulas and logic.",
    "The theorem explains the relationship between sides of a triangle.",
    "Calculus is used to study change and motion.",
    "History studies past events civilizations and empires.",
    "Ancient empires expanded through wars and trade.",
    "Historical records help people understand old societies.",
    "The king ruled the empire for many years.",
    "Archaeology helps historians learn about ancient cultures."
]

# Process input text
def process(input_text):
    tokenizer = RegexpTokenizer(r'\w+')
    stemmer = SnowballStemmer('english')
    stop_words = stopwords.words('english')

    tokens = tokenizer.tokenize(input_text.lower())
    tokens = [x for x in tokens if x not in stop_words]
    tokens_stemmed = [stemmer.stem(x) for x in tokens]

    return tokens_stemmed

# Tokenize all documents
tokens = [process(x) for x in data]

# Create dictionary
dict_tokens = corpora.Dictionary(tokens)

# Create document-term matrix
doc_term_mat = [dict_tokens.doc2bow(token) for token in tokens]

# Define number of topics
num_topics = 2

# Generate LDA model
ldamodel = models.ldamodel.LdaModel(
    doc_term_mat,
    num_topics=num_topics,
    id2word=dict_tokens,
    passes=25,
    random_state=5
)

# Print top contributing words for each topic
num_words = 5

print("Top", num_words, "contributing words to each topic:")

for item in ldamodel.print_topics(num_topics=num_topics, num_words=num_words):
    print("\nTopic", item[0])

    list_of_strings = item[1].split(" + ")
    for text in list_of_strings:
        weight = text.split("*")[0]
        word = text.split("*")[1]
        print(word, "-->", str(round(float(weight) * 100, 2)) + "%")
```
