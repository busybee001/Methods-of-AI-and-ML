```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Generate sample monthly time-series data
dates = pd.date_range('2000-01-01', periods=200, freq='M')
values = np.random.randint(10, 100, size=200)

series = pd.Series(values, index=dates)

# Basic statistics
print("Mean:", series.mean())
print("Maximum:", series.max())
print("Minimum:", series.min())
print("Standard Deviation:", series.std())

# Rolling average
rolling_mean = series.rolling(window=12).mean()

# Plot original data
plt.figure(figsize=(10,5))
plt.plot(series, label='Original Data')
plt.title("Original Time-Series Data")
plt.xlabel("Year")
plt.ylabel("Value")
plt.legend()
plt.show()

# Plot rolling average
plt.figure(figsize=(10,5))
plt.plot(series, label='Original Data')
plt.plot(rolling_mean, linewidth=3, label='12-Month Rolling Average')
plt.title("Rolling Average of Time-Series Data")
plt.xlabel("Year")
plt.ylabel("Value")
plt.legend()
plt.show()
```
