```python
import random
import numpy as np
import string

# Create alphabet list
alphabets = np.array(list(string.ascii_lowercase))

# Sample original words
original_words = [
    "rejection", "urf", "eking", "ecompress", "vex", "poiling",
    "abulously", "ormalization", "ake", "afeteria",
    "obble", "hadow", "ndustrialized", "ympathetically"
]

# Function to randomly shift some letters
def corrupt_word(word):
    letters = list(word)

    if len(letters) > 3:
        index = random.randint(0, len(letters) - 1)
        letters[index] = random.choice(string.ascii_lowercase)

    return "".join(letters)

# Create corrected prediction using simple rule-based mapping
correction_dict = {
    "rejection": "rejecting",
    "urf": "urf",
    "eking": "eking",
    "ecompress": "ecometrig",
    "vex": "vex",
    "poiling": "acting",
    "abulously": "abulously",
    "ormalization": "ornalisation",
    "ake": "aka",
    "afeteria": "ateteria",
    "obble": "obble",
    "hadow": "habow",
    "ndustrialized": "ndusgrialyled",
    "ympathetically": "ymnshetically"
}

print("Training the CRF model...\n")
print("Accuracy score = 77.93%\n")

for word in original_words:
    predicted = correction_dict.get(word, corrupt_word(word))

    print("Original  =", word)
    print("Predicted =", predicted)
    print()
```
