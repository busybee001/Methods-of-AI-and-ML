```python
import nltk

nltk.download('wordnet')
nltk.download('omw-1.4')

from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

words = [
    "wolves",
    "cars",
    "men",
    "running",
    "playing",
    "better",
    "dreamt",
    "children",
    "mice",
    "feet"
]

print("Original Word --> Lemmatized Word\n")

for word in words:
    print(f"{word} --> {lemmatizer.lemmatize(word)}")
```

```python
# Lemmatization using POS tags

print("\nVerb Lemmatization:\n")

verbs = ["running", "playing", "driving", "writing"]

for verb in verbs:
    print(f"{verb} --> {lemmatizer.lemmatize(verb, pos='v')}")
```
