```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.fft import fft, fftfreq

# Generate audio signal
sampling_freq = 44100
duration = 2
frequency = 1000

time = np.linspace(0, duration, int(sampling_freq * duration), endpoint=False)

# Create signal
audio_signal = np.sin(2 * np.pi * frequency * time)

# Take FFT
fft_signal = fft(audio_signal)
freq_axis = fftfreq(len(audio_signal), 1 / sampling_freq)

# Use only positive frequencies
positive_freqs = freq_axis[:len(freq_axis)//2]
positive_fft = np.abs(fft_signal[:len(fft_signal)//2])

# Plot frequency-domain signal
plt.figure(figsize=(10, 5))
plt.plot(positive_freqs, positive_fft, color="black")
plt.xlim(0, 3000)
plt.title("Frequency Domain Representation")
plt.xlabel("Frequency (Hz)")
plt.ylabel("Magnitude")
plt.show()
```
