```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.io import wavfile

# Generate sample audio signal
duration = 3
sampling_freq = 44100
frequency = 440

time = np.linspace(0, duration, int(sampling_freq * duration))
audio_signal = np.sin(2 * np.pi * frequency * time)

# Save audio signal as WAV file
wavfile.write(
    "sample_audio.wav",
    sampling_freq,
    (audio_signal * 32767).astype(np.int16)
)

# Read WAV file
sampling_freq, signal = wavfile.read("sample_audio.wav")

print("Sampling Frequency:", sampling_freq)
print("Audio Signal Shape:", signal.shape)

# Normalize signal
signal = signal / np.power(2, 15)

# Extract first 100 samples
signal = signal[:100]

# Create time axis in milliseconds
time_axis = 1000 * np.arange(0, len(signal), 1) / float(sampling_freq)

# Plot audio signal
plt.figure(figsize=(10, 5))
plt.plot(time_axis, signal, color="black")
plt.xlabel("Time (milliseconds)")
plt.ylabel("Amplitude")
plt.title("Audio Signal")
plt.show()
```
