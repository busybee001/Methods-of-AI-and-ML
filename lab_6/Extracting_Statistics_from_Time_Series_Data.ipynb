```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Create sample time-series data
dates = pd.date_range(start='2000-01-01', periods=150, freq='M')

x1 = pd.Series(
    np.random.normal(100, 20, 150),
    index=dates
)

x2 = pd.Series(
    np.random.normal(120, 25, 150),
    index=dates
)

# Create dataframe
data = pd.DataFrame({
    'dim1': x1,
    'dim2': x2
})

# Maximum values
print("\nMaximum values:")
print(data.max())

# Minimum values
print("\nMinimum values:")
print(data.min())

# Overall mean
print("\nOverall mean:")
print(data.mean())

# Row-wise mean (first 12 rows)
print("\nRow-wise mean (first 12 rows):")
print(data.mean(axis=1)[:12])

# Rolling mean
data.rolling(
    center=False,
    window=24
).mean().plot()

plt.title("Rolling Mean")
plt.show()

# Correlation matrix
print("\nCorrelation coefficients:")
print(data.corr())

# Rolling correlation
plt.figure()

data['dim1'].rolling(
    window=60
).corr(
    other=data['dim2']
).plot()

plt.title("Rolling Correlation")
plt.show()
```
