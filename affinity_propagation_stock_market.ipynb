```python
import numpy as np
from sklearn import covariance, cluster
```

```python
names = np.array([
    "Apple", "Microsoft", "Amazon", "Google", "Meta",
    "Tesla", "Ford", "General Motors", "Toyota", "Honda",
    "JPMorgan", "Bank of America", "Wells Fargo", "Citigroup",
    "Walmart", "Target", "Costco", "Home Depot"
])
```

```python
np.random.seed(42)

num_stocks = len(names)
num_days = 200

# Create artificial daily stock movements
base_market_movement = np.random.normal(0, 1, num_days)

delta_quotes = []

for i in range(num_stocks):
    stock_noise = np.random.normal(0, 0.5, num_days)
    stock_movement = base_market_movement + stock_noise
    delta_quotes.append(stock_movement)

delta_quotes = np.array(delta_quotes)

print("Stock movement data shape:", delta_quotes.shape)
```

```python
X = delta_quotes.copy().T

X = X / X.std(axis=0)

print("Standardized data shape:", X.shape)
```

```python
edge_model = covariance.GraphicalLassoCV()

with np.errstate(invalid="ignore"):
    edge_model.fit(X)

print("Covariance matrix shape:", edge_model.covariance_.shape)
```

```python
_, labels = cluster.affinity_propagation(edge_model.covariance_)

num_labels = labels.max()

print("Number of clusters:", num_labels + 1)
```

```python
for i in range(num_labels + 1):
    print("Cluster", i + 1, "-->", ", ".join(names[labels == i]))
```
