```python
import random

particles = [random.uniform(-10, 10) for _ in range(5)]

def objective(x):
    return x**2

print("Initial Particles:")

for p in particles:
    print(round(p, 2))

best = min(particles, key=objective)

print("\nBest Particle:")
print(round(best, 2))

print("Objective Value:")
print(round(objective(best), 2))
```
