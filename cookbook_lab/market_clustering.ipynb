```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
```

```python
# Sample customer data
data = {
    "age": [22, 25, 47, 52, 46, 56, 23, 30, 40, 60],
    "income": [25000, 30000, 80000, 90000, 75000, 95000, 28000, 35000, 65000, 100000],
    "spending_score": [70, 65, 30, 25, 35, 20, 75, 60, 40, 15]
}

df = pd.DataFrame(data)
df
```

```python
# Scale the data
scaler = StandardScaler()
scaled_data = scaler.fit_transform(df)

scaled_data
```

```python
# Apply K-Means clustering
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
df["cluster"] = kmeans.fit_predict(scaled_data)

df
```

```python
# Visualize clusters
plt.scatter(df["income"], df["spending_score"], c=df["cluster"])
plt.xlabel("Income")
plt.ylabel("Spending Score")
plt.title("Customer Market Segments")
plt.show()
```
