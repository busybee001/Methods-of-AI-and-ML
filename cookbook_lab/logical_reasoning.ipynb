```python
facts = {
    "Socrates": "man"
}

rules = [
    ("man", "mortal")
]

# Inference function
def infer(facts, rules):
    conclusions = {}

    for person, category in facts.items():
        for condition, result in rules:
            if category == condition:
                conclusions[person] = result

    return conclusions


# Run inference
conclusions = infer(facts, rules)

print("Known Facts:")
for person, category in facts.items():
    print(person, "is a", category)

print("\nApplied Rule:")
print("If someone is a man, then he is mortal.")

print("\nConclusion:")
for person, result in conclusions.items():
    print(person, "is", result)
```
