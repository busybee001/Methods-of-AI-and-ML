```python
import pandas as pd
from sklearn.metrics.pairwise import cosine_similarity
```

```python
# Sample user-item rating data
ratings = pd.DataFrame({
    "Movie_A": [5, 4, 0, 0],
    "Movie_B": [4, 5, 0, 1],
    "Movie_C": [0, 0, 5, 4],
    "Movie_D": [0, 1, 4, 5]
}, index=["User_1", "User_2", "User_3", "User_4"])

ratings
```

```python
# Calculate similarity between users
similarity = cosine_similarity(ratings)

similarity_df = pd.DataFrame(
    similarity,
    index=ratings.index,
    columns=ratings.index
)

similarity_df
```

```python
# Find users most similar to User_1
similarity_df["User_1"].sort_values(ascending=False)
```
