```python
import random

ads = ["A", "B", "C"]

clicks = {
    "A": 0,
    "B": 0,
    "C": 0
}

for i in range(20):
    ad = random.choice(ads)

    reward = random.choice([0, 1])

    clicks[ad] += reward

print("Advertisement Results")
print(clicks)

best_ad = max(clicks, key=clicks.get)

print("\nBest Advertisement:")
print(best_ad)
```
