```python
import random

N = 8

def fitness(board):
    attacks = 0

    for i in range(N):
        for j in range(i + 1, N):
            if board[i] == board[j]:
                attacks += 1

            if abs(board[i] - board[j]) == abs(i - j):
                attacks += 1

    return attacks

board = [random.randint(0, N - 1) for _ in range(N)]

print("Board:")
print(board)

print("\nFitness Score:")
print(fitness(board))
```
