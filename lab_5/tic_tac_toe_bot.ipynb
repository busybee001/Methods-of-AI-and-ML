```python
from easyAI import TwoPlayerGame, Human_Player, AI_Player, Negamax
```

```python
class TicTacToe(TwoPlayerGame):

    def __init__(self, players):
        self.players = players
        self.current_player = 1

        self.board = [
            [0, 0, 0],
            [0, 0, 0],
            [0, 0, 0]
        ]

    def possible_moves(self):
        moves = []

        for i in range(3):
            for j in range(3):
                if self.board[i][j] == 0:
                    moves.append((i, j))

        return moves

    def make_move(self, move):
        i, j = move
        self.board[i][j] = self.current_player

    def lose(self):

        lines = []

        lines.extend(self.board)

        lines.extend([
            [self.board[0][i], self.board[1][i], self.board[2][i]]
            for i in range(3)
        ])

        lines.append([
            self.board[0][0],
            self.board[1][1],
            self.board[2][2]
        ])

        lines.append([
            self.board[0][2],
            self.board[1][1],
            self.board[2][0]
        ])

        return [self.opponent_index] * 3 in lines

    def is_over(self):
        return self.lose() or len(self.possible_moves()) == 0

    def scoring(self):
        return -100 if self.lose() else 0

    def show(self):
        print("\nBoard:")
        for row in self.board:
            print(row)
```

```python
ai = Negamax(6)

game = TicTacToe([
    AI_Player(ai),
    AI_Player(ai)
])

game.play()
```
