```python
from easyAI import TwoPlayerGame, Human_Player, AI_Player, Negamax
```

```python
class ConnectFour(TwoPlayerGame):

    def __init__(self, players):
        self.players = players
        self.current_player = 1
        self.board = [[0 for _ in range(7)] for _ in range(6)]

    def possible_moves(self):
        return [str(col) for col in range(7) if self.board[0][col] == 0]

    def make_move(self, move):
        col = int(move)
        for row in range(5, -1, -1):
            if self.board[row][col] == 0:
                self.board[row][col] = self.current_player
                break

    def lose(self):
        opponent = self.opponent_index

        # horizontal
        for row in range(6):
            for col in range(4):
                if all(self.board[row][col+i] == opponent for i in range(4)):
                    return True

        # vertical
        for row in range(3):
            for col in range(7):
                if all(self.board[row+i][col] == opponent for i in range(4)):
                    return True

        # diagonal down-right
        for row in range(3):
            for col in range(4):
                if all(self.board[row+i][col+i] == opponent for i in range(4)):
                    return True

        # diagonal down-left
        for row in range(3):
            for col in range(3, 7):
                if all(self.board[row+i][col-i] == opponent for i in range(4)):
                    return True

        return False

    def is_over(self):
        return self.lose() or len(self.possible_moves()) == 0

    def scoring(self):
        return -100 if self.lose() else 0

    def show(self):
        print("\nBoard:")
        for row in self.board:
            print(row)
```

```python
ai = Negamax(5)

game = ConnectFour([
    AI_Player(ai),
    AI_Player(ai)
])

game.play()
```
